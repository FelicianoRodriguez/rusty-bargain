```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

# Datos de ejemplo (variables categóricas)
X = pd.DataFrame({
    'Ciudad': ['Madrid', 'París', 'Madrid', 'Tokio'],
    'Tamaño': ['S', 'M', 'L', 'S']
})
display(X)

# 1. Instanciar el encoder
# sparse_output=False devuelve un array normal de NumPy (en versiones antiguas era sparse=False)
encoder = OneHotEncoder(sparse_output=False)

# 2. Ajustar y transformar los datos
X_encoded = encoder.fit_transform(X)
print(X_encoded)
print(X_encoded.dtype)

# 3. Obtener el nombre de las nuevas columnas generadas
columnas_nuevas = encoder.get_feature_names_out(['Ciudad', 'Tamaño'])
print(columnas_nuevas)

# Convertir el resultado a DataFrame para visualizarlo claramente
df_encoded = pd.DataFrame(X_encoded, columns=columnas_nuevas)
print(df_encoded)
```

```python
df = pd.read_csv("datasets/car_data.csv")

# Filtros
# --------------------------------------------
# Filtrar registros con precio meno o igual a 0
filter_price = df['Price'] > 0

# Solo registros entre 1980 y 2025
filter_year = df['RegistrationYear'].between(1980, 2025)

# Solo vehículos con potencia entre 30 y 500
filter_power = df['Power'].between(30, 500)

df = df[
    filter_price &
    filter_year &
    filter_power
]

# Nullos
# -------------------------------------------
cat_columns_with_nulls = [
    'VehicleType',
    'Gearbox',
    'Model',
    'FuelType',
    'NotRepaired'
]

for column in cat_columns_with_nulls:
    df[column] = df[column].fillna('unknown')

# features y target
# ----------------------------------------------
y = df['Price']

features = [
    'VehicleType',
    'RegistrationYear',
    'Gearbox',
    'Power',
    'Model',
    'Mileage',
    'RegistrationMonth',
    'FuelType',
    'Brand',
    'NotRepaired'
]

X = df[features]

# Separar columnas numéricas y categoricas
# -----------------------------------------------------
# Esto es necesario porque cada tipo de columna requiere 
# un trantamiento distinto.

numerical_columns = [
    'RegistrationYear',
    'Power',
    'Mileage',
    'RegistrationMonth'
]

categorical_columns = [
    'VehicleType',
    'Gearbox',
    'Model',
    'FuelType',
    'Brand',
    'NotRepaired'
]

# Separar en conjunto de entrenamiento y prueba
# ----------------------------------------
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
)

encoder = OneHotEncoder(handle_unknown='ignore', sparse_output=False)

X_train_encoded = encoder.fit_transform(X_train[categorical_columns])
X_train_final = pd.concat([
    X_train[numerical_columns].reset_index(drop=True),
    pd.DataFrame(X_train_encoded, columns=encoder.get_feature_names_out(categorical_columns))
], axis=1)

X_test_encoded = encoder.transform(X_test[categorical_columns])
X_test_final = pd.concat([
    X_test[numerical_columns].reset_index(drop=True),
    pd.DataFrame(X_test_encoded, columns=encoder.get_feature_names_out(categorical_columns))
], axis=1)

display(X_train_final)
display(X_test_final)
```
